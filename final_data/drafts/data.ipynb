<h3>Data Merging - Using Python</h3>
<hr>
<br>
Here, we have several .csv files scattered across various different directories in distinct hierarchies.
Thus, we used Python to merge them into a single -> .csv file

This is acheieved by
- Declare paths in a list
- Use Pandas and Scipy -> V stack to merge them into a singular dataframe
- Export the dataframe into a .csv file.

<hr><h4><b>Breakdown File - Merging

```python
#Declaration of paths for the files to be read

paths = ["/Users/samarth/Documents/SIH 26/final_syb/Breakdown_ALL.csv",
         "/Users/samarth/Documents/SIH 26/final_syb/Breakdown.csv"]
```

```python
#Importing the required libraries

import pandas as pd
import numpy as np
from scipy.sparse import vstack

data = []

for path in paths:
    temp = pd.read_csv(path)
    data.append(temp.iloc[:, :2].values)

merged = np.vstack(data)

df_breakdown = pd.DataFrame(
    merged,
    columns=pd.read_csv(paths[0]).columns[:2]
)
df_breakdown.columns = ["Collector_Emitter_Voltage_Vce", "Leakage_Current_Ic"]

df_breakdown
```

```python
#Exporting the merged dataframe to a new CSV file

df_breakdown.to_csv("/Users/samarth/Documents/SIH 26/final_syb/dataset/Breakdown.csv", index=False)
```
